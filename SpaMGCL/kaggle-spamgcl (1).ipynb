Cell 8：生成 HLN-A1 三组诊断配置

```python
%cd /kaggle/working/SpaMGCL/SpaMGCL

import os, yaml
from copy import deepcopy

DATA_ROOT = "/kaggle/input/datasets/wuvdji/smgc-data"
BASE_CFG = "configs/hlna1_best.yaml"

with open(BASE_CFG) as f:
    base = yaml.safe_load(f)

os.makedirs("configs/generated", exist_ok=True)

def make(name, sc, snf, lam_spatial, start_epoch, epochs=50):
    cfg = deepcopy(base)

    cfg["experiment"]["name"] = name
    cfg["data"]["root"] = DATA_ROOT
    cfg["data"]["label_key"] = "Spatial_Label"
    cfg["output"] = {"root": "results"}
    cfg.pop("dir", None)

    cfg["training"]["epochs"] = epochs
    cfg["training"]["spatial_start_epoch"] = start_epoch

    cfg["model"]["fine_dim"] = 32
    cfg["model"]["cluster_temperature"] = 1.0

    cfg["loss"]["lambda_cluster"] = 0.1
    cfg["loss"]["lambda_mgcl"] = 1.0
    cfg["loss"]["lambda_spatial"] = lam_spatial

    cfg["spatial"] = {
        "enabled": bool(sc or snf),
        "consistency_weighting": bool(sc),
        "negative_filter": bool(snf),
    }

    path = f"configs/generated/{name}.yaml"
    with open(path, "w") as f:
        yaml.dump(cfg, f, default_flow_style=False, allow_unicode=True)
    print(f"✅ {path}  SC={sc} SNF={snf} λ_sp={lam_spatial} start={start_epoch}")

# core：完全关闭空间机制
make("diag_hlna1_core", sc=False, snf=False, lam_spatial=0.0, start_epoch=999)

# sp0：spatial 从第 1 轮开始记录，但 λ_spatial=0（对照组，验证 spatial 有值但不进 total）
make("diag_hlna1_sp0",  sc=False, snf=False, lam_spatial=0.0, start_epoch=1)

# sc：SC 开，λ_spatial=0.1
make("diag_hlna1_sc",   sc=True,  snf=False, lam_spatial=0.1, start_epoch=1)
```

Cell 11：自动判断是否通过

```python
import json, os

def load(name):
    with open(f"results/{name}/metrics.json") as f:
        return json.load(f)

core = load("diag_hlna1_core")["loss_history"][-1]
sp0  = load("diag_hlna1_sp0")["loss_history"][-1]
sc   = load("diag_hlna1_sc")["loss_history"][-1]

print("=" * 70)
print("第一批修复验收")
print("=" * 70)

# 验收 1：SC on 是否改变 mgcl 或 mgcl_weight_std
mgcl_core = core.get("sample_contrastive", 0)
mgcl_sc   = sc.get("sample_contrastive", 0)
std_core  = core.get("mgcl_weight_std", 0)
std_sc    = sc.get("mgcl_weight_std", 0)

sc_changed = (
    abs(mgcl_sc - mgcl_core) > 1e-4
    or abs(std_sc - std_core) > 1e-4
)
print(f"[1] SC on/off 是否改变 MGCL 输入：")
print(f"    mgcl_core={mgcl_core:.5f}  mgcl_sc={mgcl_sc:.5f}  Δ={mgcl_sc - mgcl_core:+.5f}")
print(f"    std_core ={std_core:.5f}  std_sc ={std_sc:.5f}  Δ={std_sc - std_core:+.5f}")
print(f"    结果：{'✅ 通过' if sc_changed else '❌ 未通过'}")
print()

# 验收 2：spatial 有值但 λ=0 时不影响 total
sp0_spatial = sp0.get("spatial_loss", 0)
sp0_lam     = sp0.get("lambda_spatial", -1)
print(f"[2] λ_spatial=0 时 spatial 仍能记录：")
print(f"    spatial_loss={sp0_spatial:.6f}  lambda_spatial={sp0_lam}")
print(f"    结果：{'✅ 通过' if sp0_spatial > 0 and sp0_lam == 0 else '❌ 未通过'}")
print()

# 验收 3：λ_spatial=0.1 时 total 里多出 spatial 项
sc_rec  = sc.get("reconstruction", 0)
sc_mgcl = sc.get("sample_contrastive", 0)
sc_clu  = sc.get("cluster_contrastive", 0)
sc_sp   = sc.get("spatial_loss", 0)
sc_total = sc.get("total", 0)
expected = sc_rec * 1.0 + sc_mgcl * 1.0 + sc_clu * 0.1 + sc_sp * 0.1
diff = abs(sc_total - expected)

print(f"[3] λ_spatial=0.1 时 total 是否包含 spatial 项：")
print(f"    total={sc_total:.6f}  expected≈{expected:.6f}  |diff|={diff:.6f}")
print(f"    结果：{'✅ 通过' if diff < 1e-3 else '❌ 未通过（total 公式没生效或 lambda 不对）'}")
print()

print("=" * 70)
if sc_changed and sp0_spatial > 0 and sp0_lam == 0 and diff < 1e-3:
    print("🎉 第一批修复全部通过，可以发第二批 Codex（修 SNF + SNF-only）")
else:
    print("⚠️ 有项目未通过，先别进第二批，把上面表格贴给 Codex 或我")
print("=" * 70)
```

## diag2：λ_spatial=0 的四组消融诊断

### 为什么再跑一批

上一批 diag 四组（λ_spatial=0.1）：

| name | ARI | NMI | spatial | λ_sp | SC | SNF |
|---|---|---|---|---|---|---|
| core | 0.2892 | 0.3536 | 0 | 0.0 | F | F |
| sc | 0.1987 | 0.2192 | 0.797 | 0.1 | T | F |
| snf | 0.1995 | 0.2203 | 0.796 | 0.1 | F | T |
| sc_snf | 0.1988 | 0.2191 | 0.798 | 0.1 | T | T |

三组开启空间机制的版本 ARI 全部掉到 0.199 左右，几乎完全一样。
但 SC 和 SNF 是两个完全不同的机制：
- SC 改视图权重（w_std 0.0048 → 0.0103）
- SNF 改负样本 mask（neg_count 12134772 → 12121566）

作用路径不同，结果完全一致，说明掉分不是 SC/SNF 各自的问题，
而是它们共同携带的 `λ_sp=0.1` 的 spatial_loss 项。

### 关键：λ_spatial 应该 = 0

审计报告结论：
- SC 通过 MGCL 内部视图权重生效
- SNF 通过负样本 mask 生效
- 两者都不需要 `λ_spatial > 0`

我上一批加 λ_sp=0.1 是错的。这一批改回 λ_sp=0，
只让 SC/SNF 通过 MGCL 生效，才是正确的消融设计。

### 本批目的

在 λ_sp=0 下，四组 ARI：
- 都 ≈ 0.289：SC/SNF 真实无收益
- 某组 > 0.289：那组机制真实有效
- 某组 < 0.289：那组机制真实有害

无论哪种，都能得出可解释的消融表。

### 固定参数

- seed=0, epochs=50, fine_dim=32
- t=1.0, λ_mgcl=1.0, λ_cluster=0.1
- spatial_start_epoch=1（core 用 999）
- **λ_spatial=0（本批关键）**

### 与历史结果的关系

最初那次三组消融（core 0.2894 / SC 0.2883 / SC+SNF 0.2880）
其实就是在 λ_sp=0 下跑的，但当时：
- 没有 sc_enabled / snf_enabled 日志，无法确认开关是否生效
- 单次 50 轮，噪声可能大于 0.001 的差异

本批跑在修复后的代码上，有完整诊断字段，结果更可信。

```python
%cd /kaggle/working/SpaMGCL/SpaMGCL

import os, yaml
from copy import deepcopy

DATA_ROOT = "/kaggle/input/datasets/wuvdji/smgc-data"
BASE_CFG = "configs/e185_best.yaml"   # 或 e185_ablation_core.yaml

with open(BASE_CFG) as f:
    base = yaml.safe_load(f)

os.makedirs("configs/generated", exist_ok=True)

def make(name, sc, snf, lam_spatial, start_epoch, epochs=50):
    cfg = deepcopy(base)
    cfg["experiment"]["name"] = name
    cfg["data"]["root"] = DATA_ROOT
    cfg["data"]["label_key"] = "Combined_Clusters"
    cfg["output"] = {"root": "results"}
    cfg.pop("dir", None)
    cfg["training"]["epochs"] = epochs
    cfg["training"]["spatial_start_epoch"] = start_epoch
    cfg["model"]["fine_dim"] = 128
    cfg["model"]["cluster_temperature"] = 0.3
    cfg["loss"]["lambda_cluster"] = 1.0
    cfg["loss"]["lambda_mgcl"] = 3.0
    cfg["loss"]["lambda_spatial"] = lam_spatial
    cfg["spatial"] = {
        "enabled": bool(sc or snf),
        "consistency_weighting": bool(sc),
        "negative_filter": bool(snf),
    }
    path = f"configs/generated/{name}.yaml"
    with open(path, "w") as f:
        yaml.dump(cfg, f, default_flow_style=False, allow_unicode=True)
    print(f"✅ {path}  SC={sc} SNF={snf} λ_sp={lam_spatial}")

make("diag2_e185_core",   sc=False, snf=False, lam_spatial=0.0, start_epoch=999)
make("diag2_e185_sc",     sc=True,  snf=False, lam_spatial=0.0, start_epoch=1)
make("diag2_e185_snf",    sc=False, snf=True,  lam_spatial=0.0, start_epoch=1)
make("diag2_e185_sc_snf", sc=True,  snf=True,  lam_spatial=0.0, start_epoch=1)
```

## 核心组件消融：concat_z 视角的修正解读

### 之前解读的错误

上一版 markdown 写“去掉簇对比导致表征崩塌”，这是错的。
Q_argmax 崩塌不等于表征崩塌——Q 是聚类头输出，去掉簇对比后
gradC=0，聚类头没有训练信号，Q 的输出自然退化为垃圾。

要判断表征好坏，必须看 concat_z。

### 修正后的结果

| 数据集 | 配置 | Q_argmax ARI | concat_z ARI |
|---|---|---|---|
| E18.5 | core | 0.4961 | 0.3959 |
| E18.5 | no_cluster | 0.0194 | 0.4446 |
| E18.5 | no_sample | 0.1344 | 0.2018 |
| HLN-A1 | core | 0.2892 | 0.2403 |
| HLN-A1 | no_cluster | 0.0492 | 0.2414 |
| HLN-A1 | no_sample | 0.0518 | 0.0633 |

### 两个损失的分工

- **样本对比损失**：决定表征质量。去掉后 concat_z 掉 49–74%。
- **簇对比损失**：不直接改善表征（去掉后 concat_z 持平或略好），
  但是端到端聚类头 Q 的唯一训练信号（去掉后 Q 崩塌）。

### 论文叙事

> 样本级对比损失是表征学习的决定性来源；
> 簇级对比损失不直接改善表征，但为端到端聚类头提供训练信号。
> 两者分工互补，共同决定端到端聚类性能。
