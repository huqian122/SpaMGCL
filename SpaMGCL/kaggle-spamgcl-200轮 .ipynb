第一批：HLN-A1 + E18.5（core / SC+SNF）

Cell 1：生成第一批配置

```python
%cd /kaggle/working/SpaMGCL/SpaMGCL

import os, yaml
from copy import deepcopy

os.makedirs("configs/final200", exist_ok=True)
DATA_ROOT = "/kaggle/input/datasets/wuvdji/smgc-data"

# 第一批：HLN-A1、E18.5，各跑 core 和 SC+SNF
BATCH1 = {
    "hlna1_core_200": {
        "base": "configs/hlna1_best.yaml",
        "dataset": "HLN-A1", "label_key": "Spatial_Label", "n_clusters": 10,
        "fine_dim": 32, "temp": 1.0, "mgcl": 1.0, "cluster": 0.1,
        "sc": False, "snf": False, "spk": 3,
    },
    "hlna1_scsnf_200": {
        "base": "configs/hlna1_best.yaml",
        "dataset": "HLN-A1", "label_key": "Spatial_Label", "n_clusters": 10,
        "fine_dim": 32, "temp": 1.0, "mgcl": 1.0, "cluster": 0.1,
        "sc": True, "snf": True, "spk": 3,
    },
    "e185_core_200": {
        "base": "configs/e185_best.yaml",
        "dataset": "E18.5", "label_key": "Combined_Clusters", "n_clusters": 14,
        "fine_dim": 128, "temp": 0.3, "mgcl": 3.0, "cluster": 1.0,
        "sc": False, "snf": False, "spk": 3,
    },
    "e185_scsnf_200": {
        "base": "configs/e185_best.yaml",
        "dataset": "E18.5", "label_key": "Combined_Clusters", "n_clusters": 14,
        "fine_dim": 128, "temp": 0.3, "mgcl": 3.0, "cluster": 1.0,
        "sc": True, "snf": True, "spk": 3,
    },
}

for name, s in BATCH1.items():
    with open(s["base"]) as f:
        cfg = yaml.safe_load(f)
    cfg = deepcopy(cfg)
    cfg["experiment"]["name"] = name
    cfg["experiment"]["dataset"] = s["dataset"]
    cfg["data"]["root"] = DATA_ROOT
    cfg["data"]["label_key"] = s["label_key"]
    cfg["output"] = {"root": "results"}
    cfg.pop("dir", None)
    cfg["training"]["epochs"] = 200
    cfg["training"]["spatial_start_epoch"] = 1 if (s["sc"] or s["snf"]) else 999
    cfg["model"]["fine_dim"] = s["fine_dim"]
    cfg["model"]["cluster_temperature"] = s["temp"]
    cfg["model"]["num_clusters"] = s["n_clusters"]
    cfg["clustering"]["n_clusters"] = s["n_clusters"]
    cfg["loss"]["lambda_cluster"] = s["cluster"]
    cfg["loss"]["lambda_mgcl"] = s["mgcl"]
    cfg["loss"]["lambda_spatial"] = 0.0
    cfg["spatial"] = {
        "enabled": bool(s["sc"] or s["snf"]),
        "consistency_weighting": bool(s["sc"]),
        "negative_filter": bool(s["snf"]),
    }
    cfg.setdefault("graphs", {}).setdefault("spatial", {})["k"] = s["spk"]
    path = f"configs/final200/{name}.yaml"
    with open(path, "w") as f:
        yaml.dump(cfg, f, default_flow_style=False, allow_unicode=True)
    print(f"✅ {path}  SC={s['sc']} SNF={s['snf']}")
```

Cell 2：跑第一批（4 组 200 轮）

```python
%cd /kaggle/working/SpaMGCL/SpaMGCL
import os, subprocess

NAMES = ["hlna1_core_200", "hlna1_scsnf_200", "e185_core_200", "e185_scsnf_200"]

for name in NAMES:
    cfg_path = f"configs/final200/{name}.yaml"
    res_path = f"results/{name}/metrics.json"
    if os.path.exists(res_path):
        print(f"⏭️ 已存在，跳过 {name}")
        continue
    print("=" * 70, f"\n🚀 开始 {name}")
    subprocess.run(
        ["python", "experiments/run_exp.py", "--config", cfg_path],
        cwd="/kaggle/working/SpaMGCL/SpaMGCL", check=False,
    )
```

Cell 6：第一批汇总 + 检查 + 打包

```python
%cd /kaggle/working/SpaMGCL/SpaMGCL
import os, json, pandas as pd

NAMES = ["hlna1_core_200", "hlna1_scsnf_200", "e185_core_200", "e185_scsnf_200"]

rows = []
for name in NAMES:
    p = f"results/{name}/metrics.json"
    if not os.path.exists(p):
        rows.append({"name": name, "status": "未完成"})
        continue
    with open(p) as f:
        d = json.load(f)
    rep = d.get("representation_metrics", {})
    rows.append({
        "name": name,
        "dataset": d.get("dataset"),
        "Q_ARI": round(d.get("ARI", -1), 4),
        "Q_NMI": round(d.get("NMI", -1), 4),
        "concat_z_ARI": round(rep.get("concat_z", {}).get("ARI", -1), 4),
        "concat_z_NMI": round(rep.get("concat_z", {}).get("NMI", -1), 4),
    })

df = pd.DataFrame(rows)
pd.set_option("display.width", 240)
print(df.to_string(index=False))

# 检查产物
need = ["pred_labels.npy", "gt_labels.npy", "spot_ids.npy", "coords.npy",
        "embeddings.npy", "z_mean.npy", "z_concat.npy", "sc_weights.npy",
        "snf_stats.json",
        "metrics_epoch50.json", "metrics_epoch100.json", "metrics_epoch200.json",
        "checkpoint_epoch50.pt", "checkpoint_epoch100.pt",
        "checkpoint_epoch200.pt", "checkpoint_last.pt",
        "metrics.json"]

print("\n产物检查：")
for name in NAMES:
    d = f"results/{name}"
    if not os.path.exists(d):
        print(f"❌ {name} 目录不存在"); continue
    miss = [f for f in need if not os.path.exists(os.path.join(d, f))]
    if miss:
        print(f"⚠️ {name} 缺: {miss}")
    else:
        print(f"✅ {name} 全部文件齐全")

# 打包
!zip -r /kaggle/working/final200_batch1.zip results/hlna1_core_200 results/hlna1_scsnf_200 results/e185_core_200 results/e185_scsnf_200 configs/final200/

from IPython.display import FileLink
FileLink(r'/kaggle/working/final200_batch1.zip')
```
