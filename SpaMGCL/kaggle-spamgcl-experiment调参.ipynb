| d1_dim64_t03 | 64 | 0.3 | 1.0 | 0.2088 | 0.3004 | 10.887 |
| d1_dim64_t05 | 64 | 0.5 | 1.0 | 0.1854 | 0.2843 | 10.938 |
| d1_dim64_mgcl2 | 64 | 1.0 | 2.0 | **0.2315** | 0.2985 | 10.972 |
| d1_dim64_t03_mgcl2 | 64 | 0.3 | 2.0 | 0.1967 | 0.2954 | 10.804 |

**最优配置**：fine_dim=64, cluster_temperature=1.0, lambda_mgcl=2.0
- ARI 从 baseline 的 0.1026 提升到 **0.2315**（+0.129）
- 超过 SMGC 在该数据集上的 0.2218

**关键观察**：
- `fine_dim=64` 是最大杠杆（+0.107），`fine_dim=128` 反而下降（+0.067）
- `lambda_mgcl=2.0` 单独几乎无效（+0.001），但与 `dim=64` 组合有协同效应（+0.129）
- `cluster_temperature=0.3` 单独有效（+0.089），但与 `dim=64` 叠加时收益被吸收
- `dim64 + t03 + mgcl2` 三因素叠加反而下降（0.1967），说明存在过度调节
- 所有配置的 effC ≈ 11 = 类数，聚类头未形成清晰判别边界，增益主要来自表征质量

## S2-E15 调参分析

Baseline（dim128, t0.3, mgcl3）ARI=0.3544 / NMI=0.4742。

两轮共 10 个变体，**没有一个在 ARI 上超过 baseline**。dim256 的 NMI 略优（0.4932）但 ARI 持平。

**结论**：
- dim=128 是甜点，往两边都降
- t=0.3 是甜点，降到 0.15 崩了
- mgcl=3.0 是甜点，5.0 明显下降
- spatial_k=5 反而略降
